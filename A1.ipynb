```python
# Smart Home Automation System

def fan():
    while True:
        a = input("Enter ON or OFF: ")

        if a == "ON" or a == "on":
            print("Fan is ON")
            break

        elif a == "OFF" or a == "off":
            print("Fan is OFF")
            break

        else:
            print("Invalid Choice! Please enter ON or OFF.")


while True:
    print("\n--- SMART HOME AUTOMATION ---")
    print("1. Light")
    print("2. Fan")
    print("3. Open Door")
    print("4. Close Door")
    print("5. AC Control")
    print("6. Exit")

    msg = int(input("Enter your choice: "))

    if msg == 1:
        a = input("Enter ON or OFF: ")

        if a == "ON" or a == "on":
            print("Light is ON")
        elif a == "OFF" or a == "off":
            print("Light is OFF")
        else:
            print("Invalid Input!")

    elif msg == 2:
        fan()

    elif msg == 3:
        print("Door Opened.")

    elif msg == 4:
        print("Door Closed.")

    elif msg == 5:
        temp = int(input("Enter Temperature: "))

        if temp > 30 and temp <= 50:
            print("AC turned ON.")
        elif temp >= 0 and temp <= 30:
            print("AC turned OFF.")
        else:
            print("Temperature is out of range.")

    elif msg == 6:
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
