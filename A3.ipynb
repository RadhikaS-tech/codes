```python
import heapq

graph = {}
H_val = {}

n = int(input("How many vertices you want to insert: "))

for i in range(n):
    node = input("Enter vertex: ")
    child = []

    m = int(input("Enter no. of childs: "))

    for x in range(m):
        adj = input("Enter adjacent node: ")
        child.append(adj)

    h = int(input("Enter Heuristic value for parent vertex: "))

    H_val[node] = h
    graph[node] = child


# GBFS
def gbfs(graph, H_val, start, end):
    openlist = []
    closelist = []
    parent = {}

    heapq.heappush(openlist, (H_val[start], start))

    while openlist:
        print("Openlist:", openlist)
        print("Closelist:", closelist)

        x = heapq.heappop(openlist)
        b = x[1]

        if b in closelist:
            continue

        closelist.append(b)
        print("Selected:", b)

        if b == end:
            path = []

            while b != start:
                path.append(b)
                b = parent[b]

            path.append(start)
            path.reverse()

            print("Goal found!")
            print("GBFS:", closelist)
            print("Path:", path)
            return

        for child in graph[b]:
            if child not in closelist:
                heapq.heappush(openlist, (H_val[child], child))

                if child not in parent:
                    parent[child] = b

    print("Goal not found!")


start = input("Enter Starting node: ")
end = input("Enter goal state: ")

gbfs(graph, H_val, start, end)
```
