```python
# Forward Chaining
def forward_chaining(KB, F):
    KnownFacts = F.copy()
    AddedNewFact = True

    while AddedNewFact:
        AddedNewFact = False

        for Rule in KB:
            Antecedents = Rule[0]
            Consequent = Rule[1]

            if set(Antecedents).issubset(KnownFacts):
                if Consequent not in KnownFacts:
                    KnownFacts.add(Consequent)
                    AddedNewFact = True

    return KnownFacts


# Backward Chaining
def backward_chaining(KB, KnownFacts, Goal):
    if Goal in KnownFacts:
        return True

    MatchingRules = []

    for Rule in KB:
        if Rule[1] == Goal:
            MatchingRules.append(Rule)

    if len(MatchingRules) == 0:
        return False

    for Rule in MatchingRules:
        AllAntecedentsProven = True

        for Antecedent in Rule[0]:
            if backward_chaining(KB, KnownFacts, Antecedent) == False:
                AllAntecedentsProven = False
                break

        if AllAntecedentsProven:
            KnownFacts.add(Goal)
            return True

    return False


# Knowledge Base
KB = [
    ({"car_not_starting"}, "check_fuel"),
    ({"check_fuel"}, "fuel_is_low"),
    ({"fuel_is_low"}, "add_fuel")
]

F = {"car_not_starting"}


# Main Program
while True:
    print("\n1. Forward Chaining")
    print("2. Backward Chaining")
    print("3. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        Result = forward_chaining(KB, F)

        print("\nInitial Facts:", F)
        print("Final Facts:", Result)

        FinalConclusion = None

        for Rule in KB:
            FinalConclusion = Rule[1]

        print("Final Conclusion:", FinalConclusion)

    elif choice == "2":
        Goal = input("\nEnter goal: ")
        KnownFacts = F.copy()

        if backward_chaining(KB, KnownFacts, Goal):
            print("Goal is proved.")
            print("Known Facts:", KnownFacts)
        else:
            print("Goal cannot be proved.")

    elif choice == "3":
        print("Program ended.")
        break

    else:
        print("Invalid choice.")
```
